# Remove the five redundant incomplete records

Preserve the original orders_df.



Keep the already cleaned orders_cleaned as our starting point.



Retain the most complete record for each of the five affected Order IDs.



Preserve all other records.

```python
# Step 1: Identify the five incomplete duplicate pairs

incomplete_ids = [
    70015594, 70015688, 70015782,
    70016045, 70016139
]

affected_records = orders_cleaned[
    orders_cleaned['order_id'].isin(incomplete_ids)
].copy()

# Step 2: Count non-missing values in each record

affected_records['filled_fields'] = (
    affected_records.notna().sum(axis=1)
)

# Step 3: Keep the most complete record for each Order ID

keep_indices = (
    affected_records
    .sort_values(
        ['order_id', 'filled_fields'],
        ascending=[True, False]
    )
    .drop_duplicates(subset='order_id')
    .index
)

# Step 4: Save removed records for audit purposes

removed_records = affected_records.loc[
    ~affected_records.index.isin(keep_indices)
].drop(columns='filled_fields')

# Step 5: Remove only the redundant incomplete records

orders_cleaned = orders_cleaned.drop(
    index=removed_records.index
)

# Step 6: Check the results

print("Final Orders rows:", len(orders_cleaned))
print("Incomplete duplicate records removed:", len(removed_records))
print("Exact duplicates remaining:", orders_cleaned.duplicated().sum())
print("\nRemaining affected Order IDs:")
print(
    orders_cleaned[
        orders_cleaned['order_id'].isin(incomplete_ids)
    ]['order_id'].value_counts()
)
```

# WEEK 2: BATCH DATA CLEANING



```python
# 1. Create output folder
output_folder = Path("Cleaned_Data")
output_folder.mkdir(exist_ok=True)

# 2. Connect raw DataFrames
tables = {
    "Order_Items": data["order items"].copy(),
    "Shipping": data["shipping table"].copy(),
    "Products": data["products (1)"].copy(),
    "Inventory": data["inventory"].copy()
}

# 3. Define primary keys
primary_keys = {
    "Order_Items": "order_item_id",
    "Shipping": "shipping_id",
    "Products": "product_id",
    "Inventory": "inventory_id"
}

# 4. Define numerical columns
numeric_columns = {
    "Order_Items": [
        "quantity", "unit_price",
        "discount_percent", "line_total"
    ],
    "Shipping": ["shipping_cost"],
    "Products": [
        "weight_kg", "cost_to_manufacture"
    ],
    "Inventory": [
        "stock_on_hand", "stock_reversed",
        "reorder_point", "reorder_quantity"
    ]
}
```

```python
# 5. Correct Inventory column typo
if "reoder_point" in tables["Inventory"].columns:
    tables["Inventory"] = tables["Inventory"].rename(
        columns={"reoder_point": "reorder_point"}
    )

# 6. Prepare reports
quality_report = []
missing_report = []
key_report = []
duplicate_audit = []

cleaned_tables = {}

# 7. Clean all four tables
for table_name, df in tables.items():

    print("\nCleaning:", table_name)

    original_rows = len(df)
    original_missing = int(df.isnull().sum().sum())

    # Standardize text columns
    text_cols = df.select_dtypes(
        include=["object", "string"]
    ).columns

    for col in text_cols:
        df[col] = (
            df[col]
            .astype("string")
            .str.strip()
            .replace("", pd.NA)
        )

    # Convert numerical columns
    for col in numeric_columns[table_name]:
        df[col] = pd.to_numeric(
            df[col], errors="coerce"
        )

    # Remove exact duplicates only
    duplicate_count = int(df.duplicated().sum())

    duplicates = df[df.duplicated(keep="first")].copy()

    if not duplicates.empty:
        duplicates.insert(
            0, "source_table", table_name
        )
        duplicate_audit.append(duplicates)

    df = df.drop_duplicates(keep="first").copy()

    # Check primary key quality
    key = primary_keys[table_name]

    duplicate_key_rows = int(
        df[key].duplicated(keep=False).sum()
    )

    missing_key_count = int(df[key].isna().sum())

    # Record missing values by column
    for col in df.columns:
        missing_report.append({
            "table": table_name,
            "column": col,
            "missing_values": int(df[col].isna().sum())
        })

    # Record primary key findings
    key_report.append({
        "table": table_name,
        "primary_key": key,
        "duplicate_key_rows": duplicate_key_rows,
        "missing_key_values": missing_key_count
    })

    # Check date formats without changing partial dates
    date_columns = [
        col for col in df.columns
        if "date" in col.lower()
    ]

    date_issues = 0

    for col in date_columns:
        date_values = df[col].dropna().astype("string")

        valid_dates = date_values.str.fullmatch(
            r"\d{4}-\d{2}-\d{2}"
        )

        date_issues += int((~valid_dates).sum())

    # Save cleaned table
    cleaned_tables[table_name] = df.copy()

    df.to_csv(
        output_folder / f"{table_name}_cleaned.csv",
        index=False
    )

    # Add summary
    quality_report.append({
        "table": table_name,
        "original_rows": original_rows,
        "final_rows": len(df),
        "exact_duplicates_removed": duplicate_count,
        "missing_values_before": original_missing,
        "missing_values_after": int(df.isnull().sum().sum()),
        "duplicate_primary_key_rows": duplicate_key_rows,
        "missing_primary_keys": missing_key_count,
        "date_format_issues": date_issues
    })

    print("Original rows:", original_rows)
    print("Final rows:", len(df))
    print("Exact duplicates removed:", duplicate_count)
    print("Duplicate primary key rows:", duplicate_key_rows)
    print("Missing values:", int(df.isnull().sum().sum()))

# 8. Export quality reports
pd.DataFrame(quality_report).to_csv(
    output_folder / "Data_Quality_Report.csv",
    index=False
)

pd.DataFrame(missing_report).to_csv(
    output_folder / "Missing_Values_Report.csv",
    index=False
)

pd.DataFrame(key_report).to_csv(
    output_folder / "Primary_Key_Report.csv",
    index=False
)

# 9. Export duplicate audit
if duplicate_audit:
    pd.concat(
        duplicate_audit,
        ignore_index=True
    ).to_csv(
        output_folder / "Exact_Duplicates_Audit.csv",
        index=False
    )

# 10. Cross-table relationship validation
def normalize_keys(series):
    values = series.astype("string").str.strip()
    numbers = pd.to_numeric(values, errors="coerce")

    integer_mask = (
        numbers.notna() &
        (numbers % 1 == 0)
    )

    values.loc[integer_mask] = (
        numbers.loc[integer_mask]
        .astype("Int64")
        .astype("string")
    )

    return values

relationship_checks = [
    ("Order_Items", "order_id", orders_cleaned, "order_id"),
    ("Order_Items", "product_id", cleaned_tables["Products"], "product_id"),
    ("Shipping", "order_id", orders_cleaned, "order_id"),
    ("Inventory", "product_id", cleaned_tables["Products"], "product_id")
]

relationship_report = []

for child_name, child_col, parent_df, parent_col in relationship_checks:

    child_keys = normalize_keys(
        cleaned_tables[child_name][child_col].dropna()
    )

    parent_keys = set(
        normalize_keys(parent_df[parent_col].dropna())
    )

    unmatched = int((~child_keys.isin(parent_keys)).sum())

    relationship_report.append({
        "table": child_name,
        "column": child_col,
        "unmatched_records": unmatched
    })

pd.DataFrame(relationship_report).to_csv(
    output_folder / "Relationship_Validation_Report.csv",
    index=False
)

print("\n================================")
print("BATCH CLEANING COMPLETED!")
print("================================")
print("Files saved in:", output_folder.resolve())
print("\nCleaned tables:")

for name, df in cleaned_tables.items():
    print(name, ":", df.shape)

print("\nAll reports exported successfully!")
```

```python

# Check actual matching IDs between tables

def check_relationship(child_df, child_col, parent_df, parent_col, relationship):
    child_ids = set(child_df[child_col].dropna().unique())
    parent_ids = set(parent_df[parent_col].dropna().unique())

    matched = child_ids.intersection(parent_ids)
    unmatched = child_ids.difference(parent_ids)

    print(f"\n{relationship}")
    print("Unique IDs in child table:", len(child_ids))
    print("Unique IDs in parent table:", len(parent_ids))
    print("Matching unique IDs:", len(matched))
    print("Unmatched unique IDs:", len(unmatched))
    print("Sample unmatched IDs:", list(unmatched)[:10])


check_relationship(
    order_items, "order_id",
    orders, "order_id",
    "Order Items → Orders"
)

check_relationship(
    order_items, "product_id",
    products, "product_id",
    "Order Items → Products"
)

check_relationship(
    shipping, "order_id",
    orders, "order_id",
    "Shipping → Orders"
)

check_relationship(
    inventory, "product_id",
    products, "product_id",
    "Inventory → Products"
)
```
