### Data Cleaning - Logistics Data


```python
import pandas as pd
from pathlib import Path
```

```python
# 1. Define folder paths

input_folder = Path("Raw_Data")
output_folder = Path("Cleaned_Data")
```

```python
# 2. Store cleaning reports

cleaning_reports = []
```

```python
# 3. Read and clean all CSV files
for file in input_folder.glob("*.csv"):

    print(f"\nCleaning: {file.name}")

    df = pd.read_csv(file)

    original_rows = len(df)
    original_duplicates = df.duplicated().sum()
    original_missing = df.isnull().sum().sum()

    # Standardize column names
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
        .str.replace(r"[^a-z0-9_]", "", regex=True)
    )

    # Remove extra spaces from text columns
    for col in df.select_dtypes(include="object").columns:
        df[col] = df[col].str.strip()

    # Remove completely duplicated rows
    df = df.drop_duplicates()

    # Convert date columns
    for col in df.columns:
        if "date" in col:
            df[col] = pd.to_datetime(
                df[col], errors="coerce"
            )

    # Save cleaned file
    output_file = output_folder / f"{file.stem}_cleaned.csv"
    df.to_csv(output_file, index=False)

    # Create cleaning report
    cleaning_reports.append({
        "File": file.name,
        "Original Rows": original_rows,
        "Final Rows": len(df),
        "Duplicates Removed": original_duplicates,
        "Missing Values Before": original_missing,
        "Missing Values After": df.isnull().sum().sum()
    })

    print("Completed:", file.name)
```

```python
# 4. Display cleaning report
report = pd.DataFrame(cleaning_reports)

display(report)
```

```python
from pathlib import Path

output_folder = Path("Cleaned_Data")

# Create folder if it doesn't exist
output_folder.mkdir(parents=True, exist_ok=True)

print("Cleaned_Data folder is ready!")
print("Folder location:", output_folder.resolve())
```

```python
report.to_csv(
    output_folder / "Cleaning_Report.csv",
    index=False
)

print("Cleaning report saved successfully!")
```
